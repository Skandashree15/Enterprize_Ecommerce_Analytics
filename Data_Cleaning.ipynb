```python
###   Financial Sales Dataset Cleaning and Exploration
```

```python
##  Step 1: Import Required Libraries
import pandas as pd
import numpy as np
```

```python
##  Step 2: Load the Dataset
from google.colab import files
uploaded = files.upload()
df = pd.read_excel("Sample data (1).xlsx")
```

```python
## Step 3: Explore and Clean the Dataset
df.head()
df.shape
df.describe()
df['Discount Band'] = df['Discount Band'].fillna('None')
df.columns = df.columns.str.strip()
```

```python
##  Step 4: Verify Data Cleaning and Save the Cleaned Dataset
print(df.isnull().sum())
print(df.columns)
print(df.duplicated().sum())
df.to_excel("Cleaned_Data.xlsx", index=False)
from google.colab import files
files.download("Cleaned_Data.xlsx")
```

```python
## Step 5: Exploratory Data Analysis (EDA)

print("Dataset Shape:", df.shape)

print("\nTotal Sales:", df['Sales'].sum())

print("Total Profit:", df['Profit'].sum())

print("Total Units Sold:", df['Units Sold'].sum())

print("\nNumber of Countries:", df['Country'].nunique())

print("Number of Products:", df['Product'].nunique())

print("Number of Segments:", df['Segment'].nunique())

print("\nCountry Distribution")
print(df['Country'].value_counts())

print("\nProduct Distribution")
print(df['Product'].value_counts())

print("\nSegment Distribution")
print(df['Segment'].value_counts())

print("\nDiscount Band Distribution")
print(df['Discount Band'].value_counts())

print("\nDate Range")
print(df['Date'].min())
print(df['Date'].max())
```
